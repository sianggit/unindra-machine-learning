1. Melakukan uji signifikansi asosiasi fitur kategorikal dengan **Chi-Square ($X^2$)**.
2. Membangun Pipeline terintegrasi menggunakan `ColumnTransformer`.
3. Menggabungkan `StandardScaler`, `OneHotEncoder`, dan `OrdinalEncoder` tanpa kebocoran data (*Data Leakage*).

```python
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

# Simulasi Dataset Properti (500 sampel)
np.random.seed(42)
n = 500
df_prop = pd.DataFrame({
    "luas_tanah_m2": np.random.normal(120, 30, n).clip(45, 300),
    "kamar_tidur": np.random.randint(1, 6, n),
    "lokasi": np.random.choice(["Jakarta Selatan", "Depok", "Bogor", "Bekasi"], n),
    "kondisi": np.random.choice(["Kurang", "Cukup", "Bagus", "Mewah"], n),
})
df_prop["harga_juta"] = (
    df_prop["luas_tanah_m2"] * 8.5 +
    df_prop["kamar_tidur"] * 50 +
    (df_prop["lokasi"] == "Jakarta Selatan") * 250 +
    np.random.normal(0, 50, n)
)

num_cols = ["luas_tanah_m2", "kamar_tidur"]
cat_cols = ["lokasi"]
ord_cols = {"kondisi": ["Kurang", "Cukup", "Bagus", "Mewah"]}

preprocessor = ColumnTransformer(transformers=[
    ("num", StandardScaler(), num_cols),
    ("cat", OneHotEncoder(drop="first"), cat_cols),
    ("ord", OrdinalEncoder(categories=[ord_cols["kondisi"]]), ["kondisi"])
])

pipe = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(n_estimators=50, random_state=42))
])
pipe.fit(df_prop.drop(columns=["harga_juta"]), df_prop["harga_juta"])
print("Pipeline ColumnTransformer Berhasil Dilatih!")
```
