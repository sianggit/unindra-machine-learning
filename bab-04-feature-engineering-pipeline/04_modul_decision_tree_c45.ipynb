## 1. Verifikasi Hitungan Manual Soal Ujian UPM

```python
# Kasus 14 Sampel: 9 Ya, 5 Tidak
y_sample = ["Ya"]*9 + ["Tidak"]*5
ent_total = calculate_entropy(y_sample)
print(f"Entropy Total H(S): {ent_total} (Harus ~0.9402)")

# Atribut Cuaca: Cerah (2 Ya, 3 Tidak), Mendung (4 Ya, 0 Tidak), Hujan (3 Ya, 2 Tidak)
split_cerah = ["Ya"]*2 + ["Tidak"]*3
split_mendung = ["Ya"]*4 + ["Tidak"]*0
split_hujan = ["Ya"]*3 + ["Tidak"]*2

gain_res = calculate_information_gain(y_sample, [split_cerah, split_mendung, split_hujan])
print(f"Information Gain Cuaca: {gain_res['gain']} (Harus ~0.2466)")

ratio_res = calculate_gain_ratio(y_sample, [split_cerah, split_mendung, split_hujan])
print(f"Split Info Cuaca: {ratio_res['split_info']}")
print(f"Gain Ratio C4.5 Cuaca: {ratio_res['gain_ratio']}")
```

## 2. Pelatihan Decision Tree & Visualisasi Pohon Keputusan

```python
data_path = Path.cwd().parent / "data" / "raw" / "unindra" / "modul_04_decision_tree.csv"
df_cancer = pd.read_csv(data_path)

X = df_cancer.drop(columns=["target"])
y = df_cancer["target"]

clf = DecisionTreeClassifier(max_depth=3, criterion="entropy", random_state=42)
clf.fit(X, y)

plt.figure(figsize=(16, 8))
plot_tree(clf, feature_names=X.columns.tolist(), class_names=["Ganas", "Jinak"], filled=True, rounded=True)
plt.title("Struktur Pohon Keputusan (Max Depth = 3)", fontsize=13, fontweight="bold")
plt.show()
```
