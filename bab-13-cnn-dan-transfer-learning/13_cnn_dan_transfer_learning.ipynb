```python
import numpy as np

# Simulasi Arsitektur Konvolusi 2D Sederhana
# Input Image (4x4), Kernel Filter (2x2)
image = np.array([
    [1, 2, 0, 1],
    [0, 1, 3, 2],
    [2, 0, 1, 0],
    [1, 1, 2, 3]
])
kernel = np.array([
    [1, 0],
    [0, 1]
])

def simple_conv2d(img, kern):
    h, w = img.shape
    kh, kw = kern.shape
    out_h, out_w = h - kh + 1, w - kw + 1
    feature_map = np.zeros((out_h, out_w))
    for i in range(out_h):
        for j in range(out_w):
            patch = img[i:i+kh, j:j+kw]
            feature_map[i, j] = np.sum(patch * kern)
    return feature_map

f_map = simple_conv2d(image, kernel)
print("Hasil Feature Map Konvolusi:")
print(f_map)
print("\nKonsep Transfer Learning MobileNetV2 (Bab 13):")
print("1. Bobot awal dilatih dari ImageNet (1.4 juta citra)")
print("2. Lapisan bawah mendeteksi garis & tekstur (Domain Agnostik)")
print("3. Lapisan atas diganti sesuai kasus (misal 5 jenis penyakit daun)")
```
