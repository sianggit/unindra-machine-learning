## 1. Persiapan Environment & Import Pustaka

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

## 2. Inisialisasi & Karakterisasi Dataset Alternatif

```python
# Simulasi Citra Objek Kendaraan Rendah Resolusi (Mirip CIFAR-10)
# Membuat 300 sampel gambar sintetis 32x32 piksel dengan 3 saluran warna (RGB)
np.random.seed(42)
n_images = 300

# 3 Kelas: 0: Mobil, 1: Pesawat, 2: Kapal
y_labels = np.random.choice([0, 1, 2], size=n_images)
X_images = np.zeros((n_images, 32, 32, 3), dtype=np.float32)

for i in range(n_images):
    cls = y_labels[i]
    if cls == 0:    # Mobil: dominan warna aspal dan pola horizontal
        X_images[i, :, :, 0] += np.random.uniform(0.5, 0.9) # Red car
        X_images[i, 16:28, 6:26, :] += 0.4
    elif cls == 1:  # Pesawat: latar belakang langit biru (B channel tinggi)
        X_images[i, :, :, 2] += np.random.uniform(0.6, 1.0)
        X_images[i, 12:20, :, :] += 0.5
    else:           # Kapal: latar belakang laut biru tua-kehijauan
        X_images[i, :, :, 1] += np.random.uniform(0.4, 0.8)
        X_images[i, 20:, :, :] += 0.6

X_images = np.clip(X_images + np.random.normal(0, 0.05, X_images.shape), 0.0, 1.0)
print("Bentuk Tensor Citra:", X_images.shape)
```

## 3. Implementasi Algoritma & Analisis Kinerja Model

```python
# Peragaan Operasi Konvolusi Manual (Filter Sobel untuk Edge Detection)
sobel_x = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])

# Visualisasi Sample Citra
fig, axes = plt.subplots(1, 3, figsize=(10, 4))
titles = ['Simulasi Mobil', 'Simulasi Pesawat', 'Simulasi Kapal']
for idx in range(3):
    img_sample = X_images[y_labels == idx][0]
    axes[idx].imshow(img_sample)
    axes[idx].set_title(titles[idx])
    axes[idx].axis('off')

plt.suptitle('Sampel Citra Moda Transportasi Smart City')
plt.show()

print("[INFO] Pada lingkungan dengan GPU & TensorFlow/PyTorch terpasang:")
print("Arsitektur CNN: Input(32, 32, 3) -> Conv2D(32, (3,3), relu) -> MaxPool2D(2,2) -> Conv2D(64) -> Flatten -> Dense(64) -> Dense(3, softmax)")
```
