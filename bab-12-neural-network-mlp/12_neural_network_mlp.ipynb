```python
import sys
from pathlib import Path
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "src").exists() and str(p / "src") not in sys.path:
        sys.path.append(str(p / "src"))
from unindra_modules.neural_network import neuron_forward_pass_manual

# Pembuktian Resmi Bab 12 Halaman 72
# Input x = [0.5, -0.2, 0.8], Bobot w = [0.4, 0.6, -0.3], Bias b = 0.1
calc_neuron = neuron_forward_pass_manual([0.5, -0.2, 0.8], [0.4, 0.6, -0.3], bias=0.1, activation="relu")
print("=== HASIL FORWARD PASS SATU NEURON BAB 12 ===")
print(f"Nilai Linier z: {calc_neuron['linear_sum_z']}")
print(f"Output Aktivasi a = ReLU(z): {calc_neuron['output_a']}")
print(f"Status Neuron: {calc_neuron['neuron_status']}")

# Pelatihan MLPClassifier
digits = load_digits()
X_tr, X_te, y_tr, y_te = train_test_split(digits.data, digits.target, test_size=0.25, random_state=42)
scaler = StandardScaler()
X_tr_sc = scaler.fit_transform(X_tr)
X_te_sc = scaler.transform(X_te)

mlp = MLPClassifier(hidden_layer_sizes=(64, 32), max_iter=200, random_state=42)
mlp.fit(X_tr_sc, y_tr)
print(f"Akurasi MLP pada Data Uji: {mlp.score(X_te_sc, y_te) * 100:.2f}%")
```
