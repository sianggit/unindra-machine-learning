```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

sys.path.append(str(Path.cwd().parent / "src"))
from unindra_modules.kmeans import ManualKMeans1D, elbow_method_analysis

print("Modul K-Means Unindra Berhasil Dimuat!")
```

## 1. Pembuktian Hitungan Manual UPM (Bab 10 Halaman 60)

```python
# Data resmi modul Unindra: [2, 3, 8, 9, 10] dengan Centroid Awal [2, 9]
data_upm = [2, 3, 8, 9, 10]
c_awal = [2, 9]

manual_km = ManualKMeans1D(initial_centroids=c_awal, max_iter=5)
manual_km.fit(data_upm)

for step in manual_km.history:
    print(f"=== ITERASI {step['iteration']} ===")
    print(f"Centroid Awal: {step['current_centroids']}")
    print(f"Anggota Klaster 1: {step['clusters'][0]}")
    print(f"Anggota Klaster 2: {step['clusters'][1]}")
    print(f"Centroid Baru yang Dihitung: {step['new_centroids']}")
    print(f"Konvergen: {step['converged']}\n")
```

## 2. Segmentasi Pelanggan Ritel & Elbow Method

```python
data_path = Path.cwd().parent / "data" / "raw" / "unindra" / "modul_05_kmeans_retail.csv"
df_retail = pd.read_csv(data_path)

X = df_retail[["transaction_frequency", "monetary_value_idr"]]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

df_elbow = elbow_method_analysis(X_scaled, k_range=range(1, 10))

km_final = KMeans(n_clusters=3, random_state=42, n_init=10)
df_retail["cluster"] = km_final.fit_predict(X_scaled)

plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df_retail,
    x="transaction_frequency",
    y="monetary_value_idr",
    hue="cluster",
    palette="Set1",
    s=70
)
plt.title("Hasil Segmentasi Pelanggan Ritel (k=3)", fontsize=12, fontweight="bold")
plt.xlabel("Frekuensi Transaksi")
plt.ylabel("Total Belanja (IDR)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
