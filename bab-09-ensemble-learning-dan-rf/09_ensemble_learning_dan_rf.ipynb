```python
import sys
from pathlib import Path
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "src").exists() and str(p / "src") not in sys.path:
        sys.path.append(str(p / "src"))
from unindra_modules.ensemble import calculate_oob_probability

# Pembuktian Teori OOB UPM
print("Kalkulasi Probabilitas OOB:")
print(calculate_oob_probability(n_samples=1000))

# Komparasi 3 Model Ensemble
wine = load_wine()
X_tr, X_te, y_tr, y_te = train_test_split(wine.data, wine.target, test_size=0.25, random_state=42)

rf = RandomForestClassifier(n_estimators=100, oob_score=True, random_state=42).fit(X_tr, y_tr)
ada = AdaBoostClassifier(n_estimators=100, random_state=42).fit(X_tr, y_tr)
gb = GradientBoostingClassifier(n_estimators=100, random_state=42).fit(X_tr, y_tr)

print(f"Random Forest Akurasi: {rf.score(X_te, y_te)*100:.2f}% (OOB Score: {rf.oob_score_*100:.2f}%)")
print(f"AdaBoost Akurasi: {ada.score(X_te, y_te)*100:.2f}%")
print(f"Gradient Boosting Akurasi: {gb.score(X_te, y_te)*100:.2f}%")
```
