## 1. Persiapan Environment & Import Pustaka

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score
```

## 2. Inisialisasi & Karakterisasi Dataset Alternatif

```python
# Dataset Karakteristik Musik Spotify (Simulasi)
np.random.seed(42)
n_songs = 500

# 3 Profil Lagu Utama:
# 1: EDM / Club Party (High Dance, High Energy, Low Acoustic)
# 2: Acoustic Indie / Folk (Low Dance, Low Energy, High Acoustic)
# 3: Pop Mainstream (Moderate Energy, Moderate Dance, Moderate Acoustic)

genre_type = np.random.choice([1, 2, 3], size=n_songs, p=[0.35, 0.35, 0.30])
danceability = []
energy = []
acousticness = []
tempo = []

for g in genre_type:
    if g == 1:
        danceability.append(np.random.normal(0.80, 0.08))
        energy.append(np.random.normal(0.85, 0.07))
        acousticness.append(np.random.normal(0.10, 0.05))
        tempo.append(np.random.normal(128, 5))
    elif g == 2:
        danceability.append(np.random.normal(0.35, 0.10))
        energy.append(np.random.normal(0.25, 0.08))
        acousticness.append(np.random.normal(0.85, 0.08))
        tempo.append(np.random.normal(85, 10))
    else:
        danceability.append(np.random.normal(0.60, 0.10))
        energy.append(np.random.normal(0.60, 0.10))
        acousticness.append(np.random.normal(0.40, 0.10))
        tempo.append(np.random.normal(110, 8))

df_songs = pd.DataFrame({
    'danceability': np.clip(danceability, 0, 1),
    'energy': np.clip(energy, 0, 1),
    'acousticness': np.clip(acousticness, 0, 1),
    'tempo_bpm': np.round(tempo, 1)
})

df_songs.to_csv('../../data/raw/alternatif/spotify_audio_tracks.csv', index=False)
df_songs.head()
```

## 3. Implementasi Algoritma & Analisis Kinerja Model

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_songs)

# Evaluasi Silhouette Score untuk ragam k
k_range = range(2, 7)
sil_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    sil_scores.append(silhouette_score(X_scaled, labels))

plt.figure(figsize=(7, 4))
plt.plot(k_range, sil_scores, marker='o', color='purple')
plt.title('Evaluasi Jumlah Cluster Optimal (Silhouette Analysis)')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Silhouette Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()

# Final K-Means k=3
kmeans_final = KMeans(n_clusters=3, random_state=42, n_init=10)
df_songs['cluster_kmeans'] = kmeans_final.fit_predict(X_scaled)
print("Distribusi Lagu per Cluster:\n", df_songs['cluster_kmeans'].value_counts())
```
