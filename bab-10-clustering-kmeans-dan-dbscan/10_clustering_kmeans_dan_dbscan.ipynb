```python
import sys
from pathlib import Path
import numpy as np
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "src").exists() and str(p / "src") not in sys.path:
        sys.path.append(str(p / "src"))
from unindra_modules.kmeans import ManualKMeans1D

# Pembuktian Resmi Bab 10 Halaman 60
data_soal = [2, 3, 8, 9, 10]
km_1d = ManualKMeans1D(initial_centroids=[2, 9]).fit(data_soal)
for step in km_1d.history:
    print(f"Iterasi {step['iteration']}: Centroid Awal={step['current_centroids']} -> Centroid Baru={step['new_centroids']}")

# K-Means vs DBSCAN pada Data 2D
X_synth = np.array([[1, 2], [1, 4], [1, 0], [10, 2], [10, 4], [10, 0], [50, 50]])
km = KMeans(n_clusters=2, random_state=42).fit(X_synth)
dbs = DBSCAN(eps=3, min_samples=2).fit(X_synth)

print(f"Label K-Means: {km.labels_}")
print(f"Label DBSCAN (Outlier=-1): {dbs.labels_}")
```
