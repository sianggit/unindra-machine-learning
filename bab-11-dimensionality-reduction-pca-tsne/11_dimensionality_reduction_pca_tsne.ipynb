```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "src").exists() and str(p / "src") not in sys.path:
        sys.path.append(str(p / "src"))
from unindra_modules.dim_reduction import solve_pca_2x2_manual

# Pembuktian Hitungan Manual Bab 11 Halaman 66
sigma_upm = np.array([[2.0, 0.8], [0.8, 1.0]])
res_pca = solve_pca_2x2_manual(sigma_upm)
print("=== HASIL PCA MANUAL 2x2 ===")
print(f"Lambda 1: {res_pca['lambda_1 (PC1)']} ({res_pca['percentage_PC1']})")
print(f"Lambda 2: {res_pca['lambda_2 (PC2)']} ({res_pca['percentage_PC2']})")

# Reduksi Dimensi Digits (64 dimensi -> 2 dimensi)
digits = load_digits()
pca = PCA(n_components=2)
X_pca = pca.fit_transform(digits.data)

plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=digits.target, cmap="tab10", alpha=0.6, s=15)
plt.colorbar()
plt.title("Visualisasi 2D Dataset Digits dengan PCA (Bab 11)", fontsize=12, fontweight="bold")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```
