## 1. Persiapan Environment & Import Pustaka

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_olivetti_faces
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
```

## 2. Inisialisasi & Karakterisasi Dataset Alternatif

```python
# Mengambil Dataset Wajah Olivetti (40 subjek, 10 foto per orang = 400 foto)
faces_data = fetch_olivetti_faces(shuffle=True, random_state=42)
X_faces = faces_data.data
y_faces = faces_data.target

print("Dimensi Data Citra Wajah Asli:", X_faces.shape)
print("Ukuran 1 Gambar:", 64, "x", 64, "=", 64*64, "fitur piksel.")
```

## 3. Implementasi Algoritma & Analisis Kinerja Model

```python
# 1. PCA untuk Mereduksi 4096 Dimensi menjadi 50 Principal Components
pca = PCA(n_components=50, random_state=42)
X_pca = pca.fit_transform(X_faces)

print(f"Total Variansi Terjelaskan oleh 50 Komponen: {np.sum(pca.explained_variance_ratio_)*100:.2f}%")

# 2. t-SNE untuk Visualisasi 2D Non-Linear
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
X_tsne_2d = tsne.fit_transform(X_pca)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot PCA 2D Pertama
axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=y_faces, cmap='tab20', alpha=0.7)
axes[0].set_title('Proyeksi Wajah Olivetti dengan PCA (2 Komponen Pertama)')
axes[0].set_xlabel('PC 1')
axes[0].set_ylabel('PC 2')

# Plot t-SNE 2D
axes[1].scatter(X_tsne_2d[:, 0], X_tsne_2d[:, 1], c=y_faces, cmap='tab20', alpha=0.7)
axes[1].set_title('Proyeksi Wajah Olivetti dengan t-SNE (Karakteristik Kluster Identitas)')
axes[1].set_xlabel('t-SNE 1')
axes[1].set_ylabel('t-SNE 2')

plt.tight_layout()
plt.show()
```
