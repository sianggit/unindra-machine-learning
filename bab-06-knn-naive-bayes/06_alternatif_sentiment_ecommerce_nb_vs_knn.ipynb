## 1. Persiapan Environment & Import Pustaka

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

## 2. Inisialisasi & Karakterisasi Dataset Alternatif

```python
# Dataset Ulasan Produk E-Commerce (Bahasa Indonesia)
reviews_data = [
    ("barangnya sangat bagus pengiriman cepat sekali respon penjual ramah", 1),
    ("kualitas mantap original sesuai deskripsi toko amanah bintang lima", 1),
    ("packing rapi bubble wrap tebal barang berfungsi normal makasih seller", 1),
    ("sangat memuaskan belanja disini rekomended banget gaes", 1),
    ("produk original berfungsi sangat baik pengiriman kilat", 1),
    ("kecewa berat barang rusak pecah packing asal asalan penjual slow respon", 0),
    ("barang palsu tidak sesuai gambar sangat jelek jangan beli disini", 0),
    ("pengiriman sangat lambat dua minggu baru sampai barang cacat", 0),
    ("rusak tidak bisa dipakai minta retur dipersulit parah banget", 0),
    ("kualitas buruk bahan tipis gampang sobek nyesel belanja disini", 0),
    ("bagus sekali barang mulus original fast respon recommended", 1),
    ("paket mendarat selamat mantap joss berfungsi normal", 1),
    ("pelayanan buruk admin judes barang cacat parah", 0),
    ("jelek banget barang beda sama yang dipesan mengecewakan", 0),
    ("harga murah kualitas juara pengiriman kilat jos", 1),
    ("kapok beli disini barang tidak ada gunanya langsung buang", 0),
    ("sukaaa banget sama produknya mulus tanpa lecet sedikitpun", 1),
    ("seller responsif banget barang cepet nyampe mantap pol", 1),
    ("kecewa kurir lama barang datang kondisi rusak", 0),
    ("penipu barang tidak dikirim lengkap uang tidak kembali", 0)
] * 15  # Duplikasi untuk menciptakan 300 sampel teks

df_reviews = pd.DataFrame(reviews_data, columns=['teks_ulasan', 'sentimen_positif'])
df_reviews.to_csv('../../data/raw/alternatif/ecommerce_sentiment_reviews.csv', index=False)
print("Distribusi Sentimen:", df_reviews['sentimen_positif'].value_counts().to_dict())
df_reviews.head()
```

## 3. Implementasi Algoritma & Analisis Kinerja Model

```python
X_train, X_test, y_train, y_test = train_test_split(
    df_reviews['teks_ulasan'], df_reviews['sentimen_positif'], test_size=0.25, random_state=42
)

vectorizer = TfidfVectorizer(max_features=100)
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Model 1: Multinomial Naive Bayes
nb = MultinomialNB()
nb.fit(X_train_vec, y_train)
pred_nb = nb.predict(X_test_vec)

# Model 2: KNN Classifier
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_vec, y_train)
pred_knn = knn.predict(X_test_vec)

print("=== PERFORMA MULTINOMIAL NAIVE BAYES ===")
print(classification_report(y_test, pred_nb))

print("\n=== PERFORMA K-NEAREST NEIGHBORS ===")
print(classification_report(y_test, pred_knn))
```
