```python
import sys
from pathlib import Path
import numpy as np
from sklearn.datasets import load_wine
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "src").exists() and str(p / "src") not in sys.path:
        sys.path.append(str(p / "src"))
from unindra_modules.knn import ManualKNNClassifier
from unindra_modules.naive_bayes import laplace_smoothing_prob

# Verifikasi Laplace Smoothing UPM
print(f"P('hadiah' | Spam) Laplace Smoothing: {laplace_smoothing_prob(2, 6, 10, 1.0):.4f}")

# Komparasi KNN vs Gaussian NB pada Dataset Wine
wine = load_wine()
X_tr, X_te, y_tr, y_te = train_test_split(wine.data, wine.target, test_size=0.25, random_state=42)

knn = KNeighborsClassifier(n_neighbors=5).fit(X_tr, y_tr)
gnb = GaussianNB().fit(X_tr, y_tr)

print(f"Akurasi K-NN (k=5): {accuracy_score(y_te, knn.predict(X_te))*100:.2f}%")
print(f"Akurasi Gaussian NB: {accuracy_score(y_te, gnb.predict(X_te))*100:.2f}%")
```
