1. Memahami konsep **Maximal Margin Hyperplane** ($w^T x + b = 0$).
2. Membuktikan perhitungan matematis lebar margin geometris $M = 2 / \|w\|$.
3. Membandingkan fungsi Kernel (**Linear**, **RBF/Gaussian**, **Polynomial**).
4. Melakukan penyetelan hyperparameter ($C$ dan $\gamma$) dengan `GridSearchCV`.

```python
import sys
from pathlib import Path
import numpy as np
from sklearn.svm import SVC
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.preprocessing import StandardScaler

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "src").exists() and str(p / "src") not in sys.path:
        sys.path.append(str(p / "src"))
from unindra_modules.svm import calculate_svm_margin

# Pembuktian Rumus Margin UPM: w = (2, 1)
print(calculate_svm_margin([2.0, 1.0]))

# Pelatihan SVM dengan Penyetelan GridSearch
cancer = load_breast_cancer()
scaler = StandardScaler()
X_sc = scaler.fit_transform(cancer.data)
X_tr, X_te, y_tr, y_te = train_test_split(X_sc, cancer.target, test_size=0.25, random_state=42)

param_grid = {"C": [0.1, 1, 10], "gamma": ["scale", 0.01, 0.1], "kernel": ["rbf", "linear"]}
grid = GridSearchCV(SVC(random_state=42), param_grid, cv=3)
grid.fit(X_tr, y_tr)
print(f"Hyperparameter Terbaik: {grid.best_params_}")
print(f"Akurasi Uji: {grid.score(X_te, y_te) * 100:.2f}%")
```
