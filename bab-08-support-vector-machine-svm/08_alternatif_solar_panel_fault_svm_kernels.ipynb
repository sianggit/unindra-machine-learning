## 1. Persiapan Environment & Import Pustaka

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

## 2. Inisialisasi & Karakterisasi Dataset Alternatif

```python
# Dataset Simulasi Diagnosis Panel Surya
np.random.seed(99)
n_panels = 400

irradiance = np.random.uniform(200, 1000, n_panels) # W/m2
temperature = np.random.uniform(25, 75, n_panels)    # Celcius
voltage = (irradiance * 0.04) - (temperature * 0.2) + np.random.normal(30, 2, n_panels)
current = (irradiance * 0.01) + np.random.normal(5, 0.5, n_panels)

# Status: 0 = Normal, 1 = Fault / Rusak (Drop tegangan signifikan akibat hotspot)
fault = ((voltage < 35) & (temperature > 55) | (current < 4.0)).astype(int)

df_solar = pd.DataFrame({
    'irradiance': np.round(irradiance, 1),
    'temperature_c': np.round(temperature, 1),
    'voltage_v': np.round(voltage, 2),
    'current_a': np.round(current, 2),
    'is_fault': fault
})

df_solar.to_csv('../../data/raw/alternatif/solar_panel_faults.csv', index=False)
print("Persentase Kerusakan Sel:", df_solar['is_fault'].mean() * 100, "%")
df_solar.head()
```

## 3. Implementasi Algoritma & Analisis Kinerja Model

```python
X = df_solar[['temperature_c', 'voltage_v']]
y = df_solar['is_fault']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)

# Komparasi Kernel Linear vs RBF
svm_lin = SVC(kernel='linear', C=1.0)
svm_rbf = SVC(kernel='rbf', C=10.0, gamma='scale')

svm_lin.fit(X_train_sc, y_train)
svm_rbf.fit(X_train_sc, y_train)

print("Akurasi SVM Kernel Linear:", svm_lin.score(X_test_sc, y_test))
print("Akurasi SVM Kernel RBF:", svm_rbf.score(X_test_sc, y_test))
```
