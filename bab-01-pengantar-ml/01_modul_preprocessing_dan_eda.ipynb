```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Hubungkan ke modul lokal unindra_modules
sys.path.append(str(Path.cwd().parent / "src"))
from unindra_modules.preprocessing import (
    handle_missing_values,
    detect_and_handle_outliers,
    evaluate_imbalanced_classification
)

print("Modul Preprocessing Unindra Berhasil Dimuat!")
```

## 1. Memuat Dataset Mentah Praktikum Unindra

```python
raw_data_path = Path.cwd().parent / "data" / "raw" / "unindra" / "modul_01_preprocessing.csv"
df = pd.read_csv(raw_data_path)
print(f"Dimensi Data: {df.shape[0]} baris, {df.shape[1]} kolom")
df.head()
```

## 2. Inspeksi Missing Values & Imputasi

```python
missing_info = df.isna().sum()
print("Jumlah Missing Value per Fitur:")
print(missing_info[missing_info > 0])

# Imputasi dengan KNNImputer
df_cleaned = handle_missing_values(df, strategy="knn", n_neighbors=5)
print(f"Sisa Missing Values setelah Imputasi: {df_cleaned.isna().sum().sum()}")
```

## 3. Deteksi Outlier dengan Isolation Forest

```python
df_inliers, inlier_mask = detect_and_handle_outliers(df_cleaned, contamination=0.02)
print(f"Data Asli: {len(df_cleaned)} baris")
print(f"Data Bersih (Inliers): {len(df_inliers)} baris")
print(f"Outlier Dieliminasi: {len(df_cleaned) - len(df_inliers)} baris")
```

## 4. Visualisasi Distribusi Fitur

```python
plt.figure(figsize=(10, 5))
sns.histplot(df_inliers["feature_1"], kde=True, color="#2563eb", bins=30)
plt.title("Distribusi Fitur 1 Setelah Pembersihan Data", fontsize=13, fontweight="bold")
plt.xlabel("Nilai Fitur 1")
plt.ylabel("Frekuensi")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

## 5. Bahaya Accuracy Paradox pada Imbalanced Data

```python
# Simulasi Model Bodoh (Selalu menebak kelas mayoritas 0)
y_true = df["target"].values
y_dummy = np.zeros_like(y_true)

eval_res = evaluate_imbalanced_classification(y_true, y_dummy)
print(f"Akurasi Model Bodoh: {eval_res['accuracy'] * 100:.2f}%")
print(f"F1-Score: {eval_res['f1_weighted']:.4f}")
print(f"Catatan Edukasi: {eval_res['note']}")
```
