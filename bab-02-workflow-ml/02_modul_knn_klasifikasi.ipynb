## 1. Pembuktian Hitungan Manual UPM dengan ManualKNNClassifier

```python
# Data Latih Mini UPM
X_mini = np.array([[70, 8], [60, 4], [80, 10], [50, 3]])
y_mini = np.array(["Disetujui", "Ditolak", "Disetujui", "Ditolak"])

manual_knn = ManualKNNClassifier(k=3, metric="euclidean")
manual_knn.fit(X_mini, y_mini)

# Query Ujian: (65, 7)
query = np.array([65, 7])
penjelasan = manual_knn.explain_prediction(query)

print("=== RINCIAN HITUNGAN MANUAL LEMBAR UJIAN UPM ===")
print(f"Sampel Uji: {penjelasan['query_sample']}")
print("Jarak ke Masing-Masing Data Training:")
for item in penjelasan['top_k_neighbors']:
    print(f"  - Data: {item['train_sample']} | Jarak: {item['distance']} | Kelas: {item['target_class']}")
print(f"Distribusi Suara: {penjelasan['vote_distribution']}")
print(f"Hasil Akhir Prediksi: {penjelasan['predicted_class']}")
```

## 2. Eksperimen Dataset Anggur (Wine Dataset) & Skalasi Fitur

```python
data_path = Path.cwd().parent / "data" / "raw" / "unindra" / "modul_02_knn_dataset.csv"
df_wine = pd.read_csv(data_path)
print(f"Dataset Wine: {df_wine.shape[0]} baris, {df_wine.shape[1]} kolom")
df_wine.head(3)
```

## 3. Kurva Tuning Hyperparameter k (Membuktikan Overfitting vs Underfitting)

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = df_wine.drop(columns=["target"]).values
y = df_wine["target"].values

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.25, random_state=42)

df_eval = evaluate_knn_performance(X_train, y_train, X_test, y_test, k_range=range(1, 26))

plt.figure(figsize=(10, 5))
plt.plot(df_eval["k"], df_eval["train_accuracy"], marker="o", label="Akurasi Train")
plt.plot(df_eval["k"], df_eval["test_accuracy"], marker="s", label="Akurasi Test")
plt.title("Evaluasi Kurva Akurasi Nilai k", fontsize=12, fontweight="bold")
plt.xlabel("Nilai k")
plt.ylabel("Akurasi")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()
```
