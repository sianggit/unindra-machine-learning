## 1. Persiapan Environment & Import Pustaka

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve, train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

## 2. Inisialisasi & Karakterisasi Dataset Alternatif

```python
# Simulasi Dataset HR Analytics
np.random.seed(42)
n_karyawan = 400

satisfaction = np.random.uniform(0.1, 1.0, n_karyawan)
last_evaluation = np.random.uniform(0.3, 1.0, n_karyawan)
projects_count = np.random.randint(2, 7, n_karyawan)
average_monthly_hours = np.random.normal(200, 35, n_karyawan)
tenure_years = np.random.randint(1, 10, n_karyawan)

# Resign logic: kepuasan rendah atau jam kerja ekstrem
prob_resign = (1.0 - satisfaction) * 0.5 + (average_monthly_hours > 240) * 0.3 + (projects_count > 5) * 0.2
left = (prob_resign + np.random.normal(0, 0.1, n_karyawan) > 0.55).astype(int)

df_hr = pd.DataFrame({
    'satisfaction_level': np.round(satisfaction, 2),
    'last_evaluation': np.round(last_evaluation, 2),
    'number_project': projects_count,
    'average_montly_hours': np.round(average_monthly_hours, 0),
    'time_spend_company': tenure_years,
    'left': left
})

df_hr.to_csv('../../data/raw/alternatif/hr_attrition_dataset.csv', index=False)
print("Persentase Karyawan Resign:", (df_hr['left'].mean() * 100), "%")
df_hr.head()
```

## 3. Implementasi Algoritma & Analisis Kinerja Model

```python
X = df_hr.drop('left', axis=1)
y = df_hr['left']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

depths = [1, 2, 3, 5, 8, 15, None]
train_scores = []
test_scores = []

for d in depths:
    clf = DecisionTreeClassifier(max_depth=d, random_state=42)
    clf.fit(X_train, y_train)
    train_scores.append(accuracy_score(y_train, clf.predict(X_train)))
    test_scores.append(accuracy_score(y_test, clf.predict(X_test)))

plt.figure(figsize=(9, 5))
labels = [str(d) if d else 'Unconstrained' for d in depths]
plt.plot(labels, train_scores, marker='o', label='Training Accuracy (Bias Check)')
plt.plot(labels, test_scores, marker='s', label='Testing Accuracy (Variance Check)')
plt.title('Diagnosis Bias-Variance Trade-off pada Decision Tree Depth')
plt.xlabel('Tree Max Depth')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
