```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.impute import KNNImputer
from sklearn.ensemble import IsolationForest

for p in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (p / "src").exists() and str(p / "src") not in sys.path:
        sys.path.append(str(p / "src"))
from unindra_modules.preprocessing import handle_missing_values, detect_and_handle_outliers

raw_path = Path.cwd().parent / "data" / "raw" / "unindra" / "modul_01_preprocessing.csv"
df = pd.read_csv(raw_path)
print(f"Missing Values Awal: {df.isna().sum().sum()}")

# Imputasi Cerdas
df_imputed = handle_missing_values(df, strategy="knn", n_neighbors=5)
print(f"Missing Values Pasca Imputasi: {df_imputed.isna().sum().sum()}")

# Deteksi Anomali
df_clean, mask = detect_and_handle_outliers(df_imputed, contamination=0.02)
print(f"Baris Bersih: {len(df_clean)} (Outlier disaring: {len(df_imputed) - len(df_clean)})")
```
